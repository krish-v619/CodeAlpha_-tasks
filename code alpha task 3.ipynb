```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset (required for the new tab)
df = sns.load_dataset('titanic')

# Set a clean visual theme for all plots
sns.set_theme(style="whitegrid")

# 2. Bar Chart - Survival Rate by Gender
plt.figure(figsize=(7, 5))
sns.barplot(data=df, x='sex', y='survived', hue='sex', palette='Set2', legend=False)
plt.title("Survival Rate by Gender", fontsize=14, fontweight='bold')
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Survival Rate", fontsize=12)
plt.show()

# 3. Pie Chart - Passenger Survival Distribution
plt.figure(figsize=(6, 6))
df['survived'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    labels=['Not Survived', 'Survived'],
    colors=['#ff9999','#66b3ff'],
    startangle=90
)
plt.title("Passenger Survival Distribution", fontsize=14, fontweight='bold')
plt.ylabel("")
plt.show()

# 4. Histogram - Passenger Age Distribution
plt.figure(figsize=(7, 5))
sns.histplot(data=df, x='age', bins=20, kde=True, color='teal')
plt.title("Passenger Age Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Age", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.show()

# 5. Bar Chart - Survival Rate by Passenger Class
plt.figure(figsize=(7, 5))
sns.barplot(data=df, x='pclass', y='survived', hue='pclass', palette='muted', legend=False)
plt.title("Survival Rate by Passenger Class", fontsize=14, fontweight='bold')
plt.xlabel("Passenger Class (1 = 1st, 2 = 2nd, 3 = 3rd)", fontsize=12)
plt.ylabel("Survival Rate", fontsize=12)
plt.show()

# 6. Box Plot - Age Distribution by Gender
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='sex', y='age', hue='sex', palette='Pastel1', legend=False)
plt.title("Age Distribution by Gender", fontsize=14, fontweight='bold')
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Age", fontsize=12)
plt.show()

# 7. Heatmap - Correlation Matrix
plt.figure(figsize=(8, 6))
correlation = df[['survived', 'pclass', 'age', 'fare', 'sibsp', 'parch']].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', linewidths=0.5, fmt=".2f")
plt.title("Correlation Heatmap", fontsize=14, fontweight='bold')
plt.show()

print("All visualizations completed successfully!")
```

Observaion

2. Observations and Analysis
1. Gender vs Survival (Bar Chart)
The bar chart shows the relationship between gender and passenger survival. Female passengers had a higher survival rate than male passengers. This indicates that gender played an important role in passenger survival.
2. Survival Distribution (Pie Chart)
The pie chart represents the distribution of passengers who survived and those who did not survive. It shows that more passengers died than survived during the Titanic disaster.
3. Age Distribution (Histogram)
The histogram shows the age distribution of passengers. Most passengers belonged to the young and middle-aged groups. It helps us understand the age pattern of passengers travelling on the Titanic.
4. Passenger Class vs Survival (Bar Chart)
The bar chart compares survival across different passenger classes. First-class passengers had a higher survival rate compared to second-class and third-class passengers.
5. Age Distribution by Gender (Box Plot)
The box plot compares the age distribution of male and female passengers. It helps identify differences in their median ages, age ranges and outliers.
6. Correlation Heatmap
The correlation heatmap shows relationships between numerical variables in the Titanic dataset. It helps identify positive and negative correlations between different variables.

Data Story

Passengers travelling in first class also had better survival rates than those travelling in lower classes. The age distribution shows that passengers belonged to different age groups.
