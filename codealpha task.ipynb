```python
# Import required libraries
import pandas as pd
import seaborn as sns

# 1. Load a standard sample dataset (The Titanic dataset)
df = sns.load_dataset('titanic')

print("=== 1. DATASET STRUCTURE & DATA TYPES ===")
print(df.info())

print("\n=== 2. FIRST 5 ROWS OF THE DATASET ===")
print(df.head())

print("\n=== 3. CHECKING FOR MISSING VALUES / ANOMALIES ===")
print(df.isnull().sum())

print("\n=== 4. STATISTICAL SUMMARY ===")
print(df.describe())
```

```python
print("Total Duplicate Records:", df.duplicated().sum())
```

```python
df['age'] = df['age'].fillna(df['age'].median())

df['embarked'] = df['embarked'].fillna(
    df['embarked'].mode()[0]
)

df['embark_town'] = df['embark_town'].fillna(
    df['embark_town'].mode()[0]
)

print(df.isnull().sum())
```

```python
print("\n=== 2. STATISTICAL ANALYSIS ===")

print("Average Age:", df['age'].mean())
print("Average Fare:", df['fare'].mean())
print("Highest Age:", df['age'].max())
print("Lowest Age:", df['age'].min())
print("Total Passengers:", len(df))
```

```python
print("\n=== 3. SURVIVAL ANALYSIS ===")

print("Passenger Survival Count:")
print(df['survived'].value_counts())

print("Total Survivors:", df['survived'].sum())

print("Survival Percentage:",
      df['survived'].mean() * 100)
```

```python
print("\n=== 4. GENDER-BASED SURVIVAL RATE ===")

gender_survival = df.groupby('sex')['survived'].mean() * 100

print(gender_survival)
```

```python
print("\n=== 5. PASSENGER CLASS SURVIVAL RATE ===")

class_survival = df.groupby('pclass')['survived'].mean() * 100

print(class_survival)
```

```python
print("\n=== 6. AGE ANALYSIS ===")

print("Average Age of Passengers:",
      df['age'].mean())

print("Youngest Passenger:",
      df['age'].min())

print("Oldest Passenger:",
      df['age'].max())
```

```python
print("\n=== 8. FINAL MISSING VALUE CHECK ===")

print(df.isnull().sum())
```
